```python
df['loan_status'] = df['loan_status'].astype('category')
df['loan_status'].value_counts()
```

```python
import pandas as pd
df_filtered = df[df["loan_status"].isin(["Fully Paid", "Charged Off"])].copy()
df_filtered["loan_status"] = df_filtered["loan_status"].map({"Fully Paid": 1, "Charged Off": 0})
print(df_filtered["loan_status"].value_counts())

```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set figure size
plt.figure(figsize=(8, 5))

# Create count plot
sns.countplot(data=df_filtered, x="loan_status", palette=["red", "blue"])

# Add labels and title
plt.xticks(ticks=[0, 1], labels=["Charged Off (0)", "Fully Paid (1)"])
plt.xlabel("Loan Status")
plt.ylabel("Count")
plt.title("Distribution of Loan Status (Fully Paid vs. Charged Off)")

# Show plot
plt.show()
```
